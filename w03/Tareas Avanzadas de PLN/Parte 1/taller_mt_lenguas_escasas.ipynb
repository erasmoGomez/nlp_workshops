## 🧠 Cargar modelo preentrenado multilingüe
Usaremos **MarianMT** de Helsinki-NLP y **NLLB** de Meta AI.

```python
from transformers import MarianMTModel, MarianTokenizer, pipeline

# Puedes cambiar por 'nllb' si quieres usar otro modelo
model_name = 'Helsinki-NLP/opus-mt-es-en'  # Español a Inglés como ejemplo
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

translator = pipeline('translation', model=model, tokenizer=tokenizer)
print("Modelo cargado correctamente.")
```

## 🔤 Prueba de traducción
Prueba traduciendo frases del español a otro idioma. Cambia el modelo para otras lenguas.

```python
texto = [
    "El maestro enseña en la comunidad.",
    "La niña juega cerca del río.",
    "¿Dónde está tu casa?"
]

traducciones = translator(texto)
for t in traducciones:
    print(t['translation_text'])
```
