## Paso 2: Cargar el modelo y el tokenizador

Usaremos el modelo `distilgpt2`, que es rápido y liviano.


```python

from transformers import AutoModelForCausalLM, AutoTokenizer
import torch

# Cargamos el modelo y el tokenizador
model_name = "distilgpt2"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```


---

## Paso 3: Ingresar un prompt

Puedes cambiar el contenido del prompt para probar diferentes ideas.


```python

prompt = "Había una vez un dragón que vivía en las montañas y"
input_ids = tokenizer.encode(prompt, return_tensors="pt")
```


---

## Paso 4: Generación con distintas estrategias

Probamos primero con una estrategia básica: **Greedy Decoding**.


```python
greedy_output = model.generate(input_ids, max_length=50)
print(tokenizer.decode(greedy_output[0], skip_special_tokens=True))
```


Ahora probamos con **Top-k Sampling** (aleatoriedad controlada).



topk_output = model.generate(input_ids, do_sample=True, top_k=50, max_length=50)
print("Top-k sampling:
", tokenizer.decode(topk_output[0], skip_special_tokens=True))


```python
beam_output = model.generate(input_ids, max_length=50, num_beams=5, early_stopping=True)
print(tokenizer.decode(beam_output[0], skip_special_tokens=True))
```

```python
topk_output = model.generate(input_ids, do_sample=True, top_k=50, max_length=50)
print(tokenizer.decode(topk_output[0], skip_special_tokens=True))
```

```python
topp_output = model.generate(input_ids, do_sample=True, top_p=0.92, top_k=0, max_length=50)
print(tokenizer.decode(topp_output[0], skip_special_tokens=True))
```

```python
temperature_output = model.generate(input_ids, do_sample=True, temperature=1.5, top_k=50, max_length=50)
print(tokenizer.decode(temperature_output[0], skip_special_tokens=True))
```


Y con **Top-p (Nucleus) Sampling**, otra forma de aleatoriedad.



topp_output = model.generate(input_ids, do_sample=True, top_p=0.92, top_k=0, max_length=50)
print("Top-p sampling:
", tokenizer.decode(topp_output[0], skip_special_tokens=True))
